## Step 3: Create Markdown Content from the Dataset

For each traffic violation in the dataset, you will generate markdown text that describes the violation and the associated fine. Create a loop to iterate through the dataset and store the generated markdown in a list. Each fine should look like this:

**المخالفة** - الغرامة

````python
directory = 'markdown_files'
os.makedirs(directory, exist_ok=True)
````

````python
violations = df['المخالفة']
palanties = df['الغرامة']

for i in range(len(df)):
  violation = violations.iloc[i]
  palanty = palanties.iloc[i]

  markdown_text = f'# {violation}\n'
  markdown_text += f'{palanty}'

  with open(f'{directory}/{i}.md', 'w', encoding='utf-8') as file:
    file.write(markdown_text)
````

## Step 4: Creating Documents

We will read Markdown files from a specified directory, convert their content to HTML, create Document objects, and demonstrate the process by printing the first document.

````python
marked_text = []
for filename in os.listdir(directory):
  if filename.endswith('.md'):
     with open(os.path.join(directory, filename), 'r', encoding='utf-8') as file:
      temp = file.read()
      html_text = markdown.markdown(temp)
      marked_text.append(html_text)
````

````python
marked_text[:10]
````

````python
documents = []
for text in marked_text:
    document = Document(page_content=text)
    documents.append(document)

# Print the first document
print(documents[0])
````

## Step 5: Generate Embeddings for the Documents

Generate embeddings for the chunks of text using HuggingFace's pre-trained Arabic language model. These embeddings will be stored in a `Chroma` vector store.

````python
embeddings = HuggingFaceEmbeddings(model_name="Omartificial-Intelligence-Space/GATE-AraBert-v0")
````

````python
arabic_VDB = Chroma.from_documents(documents, embeddings, persist_directory="./chroma_db")
````

# Step 6: Define the RAG Prompt Template

Define a custom prompt template in Arabic to retrieve traffic violation-related answers based on the context. Ensure the template encourages the model to give **advice** in **Arabic**, staying within the context provided.

````python
prompt = """
## التعليمات:



**المهمة:**

* أنت نظام ذكي متخصص في الاستعلام عن المخالفات المرورية في المملكة العربية السعودية.

* ستُزود بمعلومات عن المخالفة المرورية التي يرغب المستخدم في الاستعلام عنها.

* يجب عليك توفير إجابات مفصلة وشاملة بناءً على المعلومات المتاحة.

* تأكد من تقديم نصائح مفيدة باللغة العربية تتناسب مع سياق المخالفة.

* احرص على الالتزام بالمعلومات المقدمة وعدم الخروج عن سياق المخالفة.

* يجب عليك عدم الخروج عن المعلومات المسجلة في النظام.

* اذا كانت المخالفة المزوده لك ليست ذات صلة، يمكنك اخبار المستخدم بإنك لا تعرف الاجابة.

* يجب عليك ارجاع الغرامة بشكل صحيح حسب نظام المرور وحسب سؤال المستخدم.



**أسئلة الاستعلام:**

* **قيمة الغرامة:** استفسار عن قيمة الغرامة المالية المترتبة على المخالفة.



**أمثلة على صياغة الأسئلة:**

* "ما هي قيمة غرامة تجاوز السرعة بـ 30 كم/ساعة؟"



**الردود المتوقعة:**

* **معلومات دقيقة:** سيتم تقديم معلومات دقيقة عن الغرامة بناءً على قاعدة البيانات المتاحة.

* **نصائح مفيدة:** سيتم تقديم نصائح باللغة العربية حول كيفية تجنب تكرار المخالفة.

* **إجابات مفتوحة:** في حالة عدم توفر معلومات كافية، سيتم توجيه المستخدم إلى الاتصال بمركز خدمات المرور للحصول على المساعدة.



**ملاحظات هامة:**

* **الخصوصية:** يرجى العلم أن جميع المعلومات المقدمة ستعامل بسرية تامة.

* **الدقة:** تأكد من صحة المعلومات التي تقدمها للحصول على نتائج دقيقة.

قد يسأل المستخدم عن سؤال خارج نظام المرور، لذلك يمكنك الامتناع عن الاجابة (مثال: أعتذر، ولكن النظام مصمم للإجابة عن استفسارات تتعلق بالمخالفات المرورية والغرامات. )

**مثال على الحوار:**



**المستخدم:** ما هي قيمة غرامة عدم ارتداء حزام الأمان؟

**النظام:** تتراوح غرامة عدم ارتداء حزام الأمان بين 150 و300 ريال سعودي. احرص على ارتدائه دائماً. ننصحك بارتداء حزام الأمان في كل الأوقات لسلامتك وسلامة الآخرين.

المخالفة: {المخالفة}
السؤال: {السؤال}
جوابك:
"""

arabic_prompt = PromptTemplate(
    template=prompt, input_variables= ["المخالفة", "السؤال"]
)
````

## Step 7: Initialize the Language Model

Initialize the language model using the Groq API. Set up the model with a specific configuration, including the API key, temperature setting, and model name.

````python
key_API = 'API_KEY'
````

````python
LLM = ChatGroq(temperature=0, groq_api_key=key_API, model_name="llama3-70b-8192")
````

## Step 8: Create the LLM Chain

Now, you will create an LLM Chain that combines the language model and the prompt template you defined. This chain will be used to generate responses based on the retrieved context.

````python
model = LLMChain(llm=LLM, prompt=arabic_prompt, verbose=True)
````

## Step 9: Implement the Query Function

Create a function `query_rag` that will take a user query as input, retrieve relevant context from the vector store, and use the language model to generate a response based on that context.

````python
def query_rag(query: str):
    results = arabic_VDB.similarity_search_with_score(query, k=4)
    context_text = "\n\n".join([doc.page_content for doc, score in results if score > 0.5])
    rag_response = model.invoke({"المخالفة": context_text, "السؤال": query})
    return rag_response , results
````

## Step 10: Inference - Running Queries in the RAG System

In this final step, you will implement an inference pipeline to handle real-time queries. You will allow the system to retrieve the most relevant violations and fines based on a user's input and generate a response.

1. Inference Workflow:

  * The system searches for the most relevant context from the traffic violation vector store.
  * It generates an answer and advice based on the context.

# Create a Chatbot with Gradio

````python
# Function to generate answer
def generate_answer(message, history):
    result = query_rag(message)
    return result[0]['text']
````

````python
answer_bot = gr.ChatInterface(
    generate_answer,
    chatbot=gr.Chatbot(height=400),
    textbox=gr.Textbox(placeholder="ادخل استفسارك هنا", container=False, scale=7),
    title="<div style='color: #57BE6A;'>اسألنــي</div>",
    description="<div style='text-align: center;'>نظام ذكي مصمم للاستعلام عن مخالفات المرور والغرامات في المملكة العربية السعودية بناءً على لائحة الجزاءات والمخالفات المرورية</div>",
    theme="JohnSmith9982/small_and_pretty",
    examples=["ما هي غرامة وقوف المركبة في أماكن غير مخصصة للوقوف؟", "ما هي غرامة عدم ربط حزام الأمان؟", "ما هي غرامة عدم الالتزام بحدود المسارات المحددة على الطريق؟"],
    cache_examples=False,
    submit_btn="إرسال",
    stop_btn="إيقاف",
)

if __name__ == "__main__":
    answer_bot.launch(debug=True)
````
